```python
import os
import sys
import tempfile
from moviepy.editor import VideoFileClip
import whisper
import yt_dlp
import shutil
from docx import Document # Import the python-docx library


if not shutil.which("ffmpeg"):
    print("ffmpeg is not installed or not in PATH. Please install ffmpeg.")
    sys.exit(1)

def extract_audio(video_path, audio_path):
    video = VideoFileClip(video_path)
    video.audio.write_audiofile(audio_path)
    video.close()

def transcribe_audio(audio_path, model_size="base"):
    model = whisper.load_model(model_size)
    result = model.transcribe(audio_path)
    return result["text"]

def download_youtube_video(url, output_path):
    ydl_opts = {
        'format': 'bestvideo+bestaudio/best',
        'outtmpl': output_path,
        'merge_output_format': 'mp4'
    }
    try:
        print(f"Attempting to download video from: {url}")
        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            ydl.download([url])
        if not os.path.exists(output_path):
            print(f"Download failed: {output_path} not found.")
            return False
        return True
    except Exception as e:
        print(f"\nError downloading video: {e}")
        print("Please make sure you entered a valid, publicly accessible YouTube video URL.")
        return False

def save_as_docx(text, output_path):
    document = Document()
    document.add_paragraph(text)
    document.save(output_path)

def main():
    print("Video Notes Generator")
    print("1. Local Video File")
    print("2. YouTube Video URL")
    choice = input("Select input type (1/2): ").strip()
    if choice == "1":
        video_path = input("Enter path to video file: ").strip()
        if not os.path.exists(video_path):
            print("Error: Local file does not exist.")
            return
    elif choice == "2":
        url = input("Enter YouTube video URL: ").strip()
        temp_dir = tempfile.mkdtemp()
        video_path = os.path.join(temp_dir, "video.mp4")
        success = download_youtube_video(url, video_path)
        if not success:
            return
    else:
        print("Invalid choice.")
        return

    temp_audio = tempfile.mktemp(suffix=".wav")
    print("Extracting audio...")
    try:
        extract_audio(video_path, temp_audio)
    except Exception as e:
        print(f"Error extracting audio: {e}")
        return

    print("Transcribing audio...")
    notes = transcribe_audio(temp_audio)

    # Change output to docx
    output_file = "transcription.docx"
    save_as_docx(notes, output_file)

    # Clean up temporary files if they exist
    if os.path.exists(video_path) and 'tempfile' in video_path:
        shutil.rmtree(os.path.dirname(video_path))
    if os.path.exists(temp_audio):
        os.remove(temp_audio)

    print(f"Transcription saved to {output_file}")


if __name__ == "__main__":
    main()
```
